### **第一題：設計一個演算法來檢測社交網絡中的影響者**  
影響者通常具有較高的 **中心性**，可以用幾種不同的度量方式來識別：  
- **度數中心性 (Degree Centrality)**：衡量節點的直接連接數  
- **接近中心性 (Closeness Centrality)**：衡量節點與其他節點的最短距離總和  
- **中介中心性 (Betweenness Centrality)**：衡量節點出現在最短路徑中的頻率  
- **PageRank**：衡量一個節點被高影響力節點連接的程度  

#### **演算法步驟**
1. **建模社交網絡為圖** (Graph)，節點代表用戶，邊代表關係 (加權/無權)  
2. **計算中心性度量**：
   - 可以使用 **networkx** 庫來計算上述指標  
3. **選擇影響者標準**：
   - 設置門檻值，如 **PageRank > 0.05** 或 **Betweenness 排名前 10%**  
   - 或者根據 **綜合指標**（加權平均多個中心性度量）  
4. **輸出影響者列表**  

#### **Python 實現**
```python
import networkx as nx

def detect_influencers(graph, threshold=0.05):
    pagerank = nx.pagerank(graph)
    influencers = [node for node, score in pagerank.items() if score > threshold]
    return influencers

# 測試
G = nx.erdos_renyi_graph(100, 0.1)  # 隨機生成 100 個節點、連接概率 0.1 的社交網絡
influencers = detect_influencers(G)
print("Detected influencers:", influencers)
```
這種方法可以擴展為計算 **多種中心性指標** 並加權平均，提高準確度。

### **第二題：量子計算如何影響密碼學及其挑戰**
量子計算對現代密碼學的影響主要來自 **Shor’s Algorithm (對數分解)** 和 **Grover’s Algorithm (搜尋加速)**：

#### **影響**
1. **對非對稱加密 (RSA, ECC) 的威脅**
   - RSA 依賴於大數分解的計算困難性
   - Shor’s Algorithm 可以 **多項式時間內分解大數**，破壞 RSA、ECC、DSA
2. **對對稱加密 (AES) 的影響**
   - Grover’s Algorithm 可以加速 **暴力破解**，但影響較小
   - AES-256 **仍然安全**，但 AES-128 的安全性下降

#### **實施挑戰**
1. **量子硬體限制**：當前量子計算機的量子位數量遠不足以運行 Shor’s Algorithm 來破解 2048-bit RSA  
2. **糾錯需求**：量子計算易受 **退相干 (Decoherence)** 影響，需要 **糾錯碼**（如表面碼）  
3. **抗量子密碼學的發展**：標準組織 (NIST) 正在開發抗量子加密算法（如格基加密）  

**結論**：量子計算 **未來可能影響密碼學**，但短期內還無法實施大規模攻擊，**抗量子加密算法的研究正在加速**。

### **第三題：分析 Fibonacci 遞迴演算法的時間複雜度並優化**
原始程式：
```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
```
#### **時間複雜度分析**
- `T(n) = T(n-1) + T(n-2) + O(1)`
- 這是典型的 **指數時間 O(2^n) 遞迴**，因為每次遞迴都會展開兩個子問題
- 例如，`fibonacci(5)` 會展開為：
  ```
          fib(5)
         /      \
      fib(4)   fib(3)
      /   \      /   \
   fib(3) fib(2) fib(2) fib(1)
   ```
  很多子問題 **重複計算**，導致指數級別增長。

#### **改進方案：動態規劃 (DP)**
##### **方法 1：記憶化 (Top-Down DP)**
```python
def fibonacci_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci_memo(n-1, memo) + fibonacci_memo(n-2, memo)
    return memo[n]
```
**時間複雜度：O(n)，空間複雜度：O(n)（遞迴棧深度）**

##### **方法 2：迭代 (Bottom-Up DP)**
```python
def fibonacci_dp(n):
    if n <= 1:
        return n
    dp = [0, 1]
    for i in range(2, n + 1):
        dp.append(dp[i-1] + dp[i-2])
    return dp[n]
```
**時間複雜度：O(n)，空間複雜度：O(n)（可進一步優化為 O(1)）**

##### **方法 3：O(1) 空間最佳化**
```python
def fibonacci_optimized(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b
```
**時間複雜度：O(n)，空間複雜度：O(1)**

# Fibonacci Algorithm Analysis 斐波那契算法分析

### 遞迴實現分析 Recursive Implementation Analysis
* 提供的代碼使用簡單遞迴來計算斐波那契數列
  ```python
  def fibonacci(n):
      if n <= 1:
          return n
      return fibonacci(n-1) + fibonacci(n-2)
  ```
* 時間複雜度為**指數級 Exponential**：$O(2^n)$
* 空間複雜度為**線性 Linear**：$O(n)$，主要來自遞迴調用堆棧深度
* 這種實現存在大量**重複計算 Redundant Calculations**，因為相同的斐波那契值被反復計算

### 複雜度分析詳解 Complexity Analysis Explained
* 每個函數調用分裂成兩個子調用，形成二叉樹結構
* 樹的高度為$n$，最壞情況下節點數接近$2^n$
* 精確的時間複雜度為$O(1.618^n)$，即黃金比例的$n$次方
* 對於較大的$n$值，這種算法效率極低，例如$n=50$將需要數萬億次計算

### 優化解決方案：記憶化 Optimization: Memoization
* 使用**記憶化 Memoization**儲存已計算的結果以避免重複計算
  ```python
  def fibonacci_memo(n, memo={}):
      if n in memo:
          return memo[n]
      if n <= 1:
          return n
      memo[n] = fibonacci_memo(n-1, memo) + fibonacci_memo(n-2, memo)
      return memo[n]
  ```
* 時間複雜度顯著降低至**線性 Linear**：$O(n)$
* 空間複雜度仍為$O(n)$，用於儲存中間結果

### 優化解決方案：動態規劃 Optimization: Dynamic Programming
* 使用**自下而上 Bottom-up**方法的**動態規劃 Dynamic Programming**
  ```python
  def fibonacci_dp(n):
      if n <= 1:
          return n
      dp = [0] * (n + 1)
      dp[0], dp[1] = 0, 1
      for i in range(2, n + 1):
          dp[i] = dp[i-1] + dp[i-2]
      return dp[n]
  ```
* 時間複雜度：$O(n)$
* 空間複雜度：$O(n)$，儲存所有斐波那契數

### 優化解決方案：空間優化 Optimization: Space Optimization
* 只保留計算過程中的最後兩個數值
  ```python
  def fibonacci_optimized(n):
      if n <= 1:
          return n
      a, b = 0, 1
      for _ in range(2, n + 1):
          a, b = b, a + b
      return b
  ```
* 時間複雜度：$O(n)$
* 空間複雜度：$O(1)$，只使用常數空間

### 進階優化：矩陣乘法 Advanced Optimization: Matrix Exponentiation
* 使用**矩陣乘法 Matrix Exponentiation**可以將時間複雜度降至對數級
  ```python
  def fibonacci_matrix(n):
      if n <= 1:
          return n
      def matrix_multiply(A, B):
          C = [[0, 0], [0, 0]]
          for i in range(2):
              for j in range(2):
                  for k in range(2):
                      C[i][j] += A[i][k] * B[k][j]
          return C
          
      def matrix_power(A, n):
          if n <= 1:
              return A
          if n % 2 == 0:
              half_pow = matrix_power(A, n // 2)
              return matrix_multiply(half_pow, half_pow)
          else:
              return matrix_multiply(A, matrix_power(A, n-1))
              
      F = [[1, 1], [1, 0]]
      F_n = matrix_power(F, n-1)
      return F_n[0][0]
  ```
* 時間複雜度：$O(\log n)$
* 空間複雜度：$O(\log n)$，來自遞迴調用堆棧

### 各種實現性能比較 Performance Comparison
* **遞迴**：最簡單但效率最低，$n > 40$時實際運行緩慢
* **記憶化**：大幅提升效率，適用於$n$值較大的情況
* **動態規劃**：消除遞迴開銷，進一步提高效率
* **空間優化**：在保持線性時間複雜度的同時最小化內存使用
* **矩陣乘法**：理論上最快，適用於計算極大的斐波那契數

所有優化後的解決方案都能將原始遞迴算法的指數時間複雜度降低至更實用的水平，其中矩陣乘法方法在理論上效率最高。
