# HW.畫一個函數圖形

```python
# prompt: 畫一個 y=cos(x)圖形

import numpy as np
import matplotlib.pyplot as plt

# Generate x values from 0 to 2*pi
x = np.linspace(0, 2 * np.pi, 100)

# Calculate corresponding y values using the cosine function
y = np.cos(x)

# Create the plot
plt.plot(x, y)

# Add labels and title
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.title("Plot of y = cos(x)")

# Display the plot
plt.show()
```

# **畫出右半邊的愛心曲線**
## 函式庫使用
- numpy: 用於數值計算和陣列操作
- matplotlib.pyplot: 用於繪製圖形

## 資料準備
- 使用 `np.linspace(-2, 2, 400)` 建立 x, y 座標點
    - 範圍: -2 到 2
    - 取樣點數: 400 個點
- 使用 `np.meshgrid()` 建立二維網格

    - 將一維的 x, y 陣列轉換成二維網格座標系統
    - X, Y 分別代表網格上每個點的 x, y 座標值

## 方程式定義

- 原方程式: `x^2 + (y - x^(2/3))^2 = 1`
- 程式中表示為: `Z = X**2 + (Y - X**(2/3))**2 - 1`
    - `**` 表示次方運算
    - 將等式移項後，左邊為 Z，右邊為 0

## 繪圖設定
- 使用 `plt.contour()` 繪製等高線圖
    - `levels=[0]`: 只畫出 Z=0 的等高線，即原方程式的解
    - `colors='darkred'`: 設定線條顏色為深紅色
- 加入座標軸標籤和標題
    - `xlabel`: x 軸標籤
    - `ylabel`: y 軸標籤
    - `title`: 圖形標題
- 使用 `plt.grid(True)` 加入網格線增加可讀性

## 輸出結果
- `plt.show()` 顯示最終圖形
- 圖形會呈現出一個特殊的曲線，這是原方程式在 x-y 平面上的軌跡

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a grid of x and y values
x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, y)

# Define the equation
Z = X**2 + (Y - X**(2/3))**2 - 1

# Find the contour where Z is equal to 0
plt.contour(X, Y, Z, levels=[0], colors='darkred')

# Set labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("Plot of x^2 + (y - x^(2/3))^2 = 1")

# Add a grid for better readability
plt.grid(True)

# Display the plot
plt.show()
```
