
# 創意圖形：Heart Curve Drawing 心形曲線繪製

### 數學原理 Mathematical Foundation
* 程式使用參數方程式來繪製心形曲線
* 心形曲線由以下參數方程式定義：

  $$
  x = 16 \sin^3(t)
  $$

  $$
  y = 13 \cos(t) - 5 \cos(2t) - 2 \cos(3t) - \cos(4t)
  $$
#  
* 參數 `t` 從 0 到 $2\pi$ 變化，生成完整的心形

### 程式結構 Program Structure
* 導入必要的函式庫：**NumPy** 用於數學運算，**Matplotlib** 用於繪圖
* 定義 `heart_curve()` 函數來計算心形曲線的點座標
* 設定繪圖畫布大小為 8×8
* 生成 500 個 `t` 值，範圍從 0 到 $2\pi$

### 多層次效果 Layering Effect
* 使用遞減循環 `for i in range(10,0,-1)` 從大到小繪製心形
* 每層心形的比例因子從 1.0 降至 0.1，創造出層疊效果
* 使用 **Matplotlib** 的 `cm.Reds` 色彩映射，顏色隨層次變化
* 隨著心形尺寸減小，顏色由淺紅變為深紅

### 視覺美化 Visual Enhancement
* 設定 `set_aspect('equal')` 保證心形比例正確顯示
* 使用 `axis('off')` 隱藏座標軸，使圖形更加美觀
* 每條曲線的線寬設為 2，提高可視性


```python

import numpy as np
import matplotlib.pyplot as plt

# Define the heart curve equation
def heart_curve(t):
    x = 16 * np.sin(t)**3
    y = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)
    return x, y

# Create the plot
plt.figure(figsize=(8, 8))  # Adjust figure size for better visualization

# Generate t values
t = np.linspace(0, 2 * np.pi, 500)

# Iterate for different scales to create the layered effect
for i in range(10,0,-1):
    scale = i/10
    x,y = heart_curve(t)
    plt.plot(x*scale, y*scale, color = plt.cm.Reds(i/10), linewidth = 2)

plt.gca().set_aspect('equal')  # Ensure equal aspect ratio
plt.axis('off') # Hide axis
plt.show()
```
