# Pixel 風格
## Key Characteristics 主要特徵

- Limited Colors - 通常僅使用 4-16 種顏色
- Low Resolution - 典型解析度 256x224 或更低
- Sharp Pixels - 無抗鋸齒處理，保持銳利邊緣
- Grid-based - 嚴格遵循像素網格系統

## Famous Examples 著名範例

- Super Mario Bros - 任天堂經典角色設計
- Pac-Man - 簡潔的迷宮遊戲美學
- The Legend of Zelda - 奇幻冒險像素風格
